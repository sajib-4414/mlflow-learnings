### MLflow tracking server

```python
import mlflow
```

```python
mlflow.set_tracking_uri("http://127.0.0.1:5000") #insructing to tell mlflow that, any experiment that will happen, track here. and here is actually our mlflow local server running
```

```python
mlflow.set_experiment("check localhost connection")
with mlflow.start_run():
    mlflow.log_metric('test',1)
    mlflow.log_metric('krish',2)
```
